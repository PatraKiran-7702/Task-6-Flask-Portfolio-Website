```python
from flask import Flask, render_template_string, request
from threading import Thread

app = Flask(__name__)

home_page = """
<!DOCTYPE html>
<html>
<head>
    <title>My Portfolio</title>
</head>
<body style="font-family:Arial;text-align:center;">
    <h1>Kiran Brundavan</h1>
    <h2>Python Developer</h2>

    <h3>About Me</h3>
    <p>I am a CSE student interested in Python, Flask and Web Development.</p>

    <h3>Skills</h3>
    <ul style="list-style:none;">
        <li>Python</li>
        <li>Flask</li>
        <li>HTML</li>
        <li>CSS</li>
    </ul>

    <a href="/contact">Contact Me</a>
</body>
</html>
"""

contact_page = """
<!DOCTYPE html>
<html>
<head>
    <title>Contact</title>
</head>
<body style="font-family:Arial;text-align:center;">
    <h1>Contact Form</h1>

    <form method="POST">
        <input type="text" name="name" placeholder="Your Name" required><br><br>

        <input type="email" name="email" placeholder="Your Email" required><br><br>

        <textarea name="message" placeholder="Your Message" required></textarea><br><br>

        <button type="submit">Submit</button>
    </form>

    <br>
    <p>{{msg}}</p>

    <a href="/">Back to Home</a>
</body>
</html>
"""

@app.route('/')
def home():
    return render_template_string(home_page)

@app.route('/contact', methods=['GET', 'POST'])
def contact():
    msg = ""
    if request.method == 'POST':
        name = request.form['name']
        msg = f"Thank you {name}! Your message has been received."
    return render_template_string(contact_page, msg=msg)

def run():
    app.run(port=5000)

Thread(target=run).start()

print("Portfolio Website Running!")
print("Open: http://127.0.0.1:5000")
```
